**Задание 5. В двух столбцах есть пропущенные значения. Один из них — `days_employed`. Пропуски в этом столбце вы обработаете на следующем этапе. Другой столбец с пропущенными значениями — `total_income` — хранит данные о доходах. На сумму дохода сильнее всего влияет тип занятости, поэтому заполнить пропуски в этом столбце нужно медианным значением по каждому типу из столбца `income_type`. Например, у человека с типом занятости `сотрудник` пропуск в столбце `total_income` должен быть заполнен медианным доходом среди всех записей с тем же типом.**

```python
for t in data['income_type'].unique():
    data.loc[(data['income_type'] == t) & (data['total_income'].isna()), 'total_income'] = \
    data.loc[(data['income_type'] == t), 'total_income'].median()
```

### Категоризация данных

**Задание 16. На основании диапазонов, указанных ниже, создайте в датафрейме `data` столбец `total_income_category` с категориями:**

- 0–30000 — `'E'`;
- 30001–50000 — `'D'`;
- 50001–200000 — `'C'`;
- 200001–1000000 — `'B'`;
- 1000001 и выше — `'A'`.


**Например, кредитополучателю с доходом 25000 нужно назначить категорию `'E'`, а клиенту, получающему 235000, — `'B'`. Используйте собственную функцию с именем `categorize_income()` и метод `apply()`.**

```python
def categorize_income(income):
    try:
        if 0 <= income <= 30000:
            return 'E'
        elif 30001 <= income <= 50000:
            return 'D'
        elif 50001 <= income <= 200000:
            return 'C'
        elif 200001 <= income <= 1000000:
            return 'B'
        elif income >= 1000001:
            return 'A'
    except:
        pass
```

```python
data['total_income_category'] = data['total_income'].apply(categorize_income)
```

**Используйте собственную функцию с именем `categorize_purpose()` и метод `apply()`. Изучите данные в столбце `purpose` и определите, какие подстроки помогут вам правильно определить категорию.**

def categorize_purpose(row):

### Есть ли зависимость между количеством детей и возвратом кредита в срок?

```python
data_group = data.groupby('children').agg({'debt': ['count','sum']})# сгруппировали данные по столбцу children, для столбца debt применили функции count и sum
data_group
```

выяснили:
- количество клиентов по группировке кол-ву детей с учетом показателя 'debt' - столбец count.
- количество клиентов, имеющих задолженность - столбец sum.

```python
data_group['daily_debt'] = data_group['debt']['sum'] / data_group['debt']['count'] * 100 # разделили сумму должников на общее кол-во клиентов каждой категории кол-ва детей в семье
data_group
```

- посчитали долю должников в процентах - столбец daily_debt.

```python
data_group[data_group['debt']['count'] >= 300].sort_values(by='daily_debt') # удалили выделяющиеся значения и отсортировали по возрастанию данные по долевому показателю
```

- убрали выделяющиеся значения - значения для клиентов с 4-мя и 5-ю детьми.
- отсортировали таблицу(массив) по столбцу daily_debt по возрастанию.

**Вывод:** 
В данной выборке есть значения, которые сильно выделяются среди остальных - количество клиентов с 4-мя и 5-ю детьми - минимальны. Такие данные искажают результат, поэтому удалили их из таблицы.
Клиенты без детей более надежные заемщики. Клиенты с детьми, а в основной массе те, кто с 1-им или 2-мя детьми, проблемные заемщики.

### Есть ли зависимость между семейным положением и возвратом кредита в срок?

```python
def debt_categ (row): #создали новую функцию, которая принимает строку датафрейма как аргумент row
    if row == 1:
        return '1'
    if row == 0:
        return '0'
data['debt_categ'] = data['debt'].apply(debt_categ)
data['debt_categ']
```

- создали столбец debt_categ аналогичный debt для создания сводной таблицы

```python
data_pivot_family = data.pivot_table(index='family_status', columns='debt_categ', values='debt', aggfunc='count')#создали сводную таблицу
data_pivot_family
```

- создали сводную таблицу с данными по количеству клиентов с задолженностью и без задолженности по семейному положению

```python
data_pivot_family['rate'] = data_pivot_family['1'] / (data_pivot_family['1'] + data_pivot_family['0']) * 100 # создали новый столбец, показывающий долю должников по семейному положению
data_pivot_family
```

- посчитали долю должников в процентах и добавили столбец в сводную таблицу data_pivot_family

```python
data_pivot_family.sort_values(by='rate') # отсортировали по возрастанию таблицу по долевому показателю
```

- отсортировали таблицу(массив) по столбцу rate по возрастанию.

**Вывод:** 
Самый высокий процент должников приходится на людей, не состоящих в отношениях. Наблюдается прямая зависимость наличия у заемщика ярко выраженного чувства ответственности, если заемщик был или находится в зарегистрированном браке. Люди с опытом более организованные/системные - надежные заемщики.
